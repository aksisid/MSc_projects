#### Function to reverse BWT transformation

```python
def bwt_reverse(bwt):
    #insert a Burrows-Wheeler Transformation to get the original string
    
    table = [""] * len(bwt) # Initialize a table with empty strings
    for _ in range(len(bwt)): # for loop but we do not need the variable (used '_')
        table = sorted([bwt[i] + table[i] for i in range(len(bwt))])  # add  bwt characters to the beginning of table and sort
    for row in table:
        if row.endswith("$"): # find the row that ends with $ (original string)
            return row.rstrip("$")  # Remove the terminal symbol
    return None
```

```python
original_seq = bwt_reverse(received_bwt)
print("Reconstructed sequence:", original_seq)
```
